```python
# Calculate rating count and average rating

movie_popularity = ratings_clean.groupby('movieId').agg(
    average_rating=('rating', 'mean'),
    rating_count=('rating', 'count')
).reset_index()

# Add movie titles

movie_popularity = movie_popularity.merge(
    movies_clean[['movieId', 'title']],
    on='movieId'
)

print("Movie Popularity:")
display(movie_popularity.head(10))
```

```python
top_popular_movies = movie_popularity.sort_values(
    by='rating_count',
    ascending=False
).head(10)

print("Top 10 Most Popular Movies:")
display(top_popular_movies[['title', 'rating_count', 'average_rating']])
```

```python
popular_movies_100 = movie_popularity[
    movie_popularity['rating_count'] >= 100
]

print("Movies with at Least 100 Ratings:",
      len(popular_movies_100))
```

```python
# Calculate the overall average rating

C = movie_popularity['average_rating'].mean()

# Minimum number of ratings required

m = 50

# Calculate weighted rating

movie_popularity['weighted_score'] = (
    (movie_popularity['rating_count'] /
     (movie_popularity['rating_count'] + m))
    * movie_popularity['average_rating']
    +
    (m / (movie_popularity['rating_count'] + m)) * C
)

print("Weighted Movie Scores:")
display(movie_popularity.head(10))
```

```python
top_weighted_movies = movie_popularity.sort_values(
    by='weighted_score',
    ascending=False
).head(10)

print("Top 10 Movies by Weighted Rating:")
display(top_weighted_movies[
    ['title', 'average_rating', 'rating_count', 'weighted_score']
])
```

```python
from sklearn.metrics.pairwise import cosine_similarity

print("Recommendation library imported successfully!")
```

```python
user_movie_matrix = ratings_clean.pivot_table(
    index='userId',
    columns='movieId',
    values='rating'
).fillna(0)

print("User-Movie Matrix Shape:", user_movie_matrix.shape)

display(user_movie_matrix.head())
```

```python
user_similarity = cosine_similarity(user_movie_matrix)

user_similarity_df = pd.DataFrame(
    user_similarity,
    index=user_movie_matrix.index,
    columns=user_movie_matrix.index
)

print("User Similarity Matrix Shape:", user_similarity_df.shape)

display(user_similarity_df.iloc[:5, :5])
```

```python
target_user_id = 414

similar_users = user_similarity_df[target_user_id].drop(
    target_user_id
).sort_values(ascending=False)

top_similar_users = similar_users.head(10)

print("Top 10 Similar Users to User", target_user_id)
print(top_similar_users)
```

```python
# Keep only users with positive similarity

similar_users = similar_users[similar_users > 0].head(10)

# Get ratings given by similar users

neighbor_ratings = ratings_clean[
    ratings_clean['userId'].isin(similar_users.index)
].copy()

# Add similarity scores

neighbor_ratings['similarity'] = neighbor_ratings['userId'].map(
    similar_users
)

# Calculate weighted ratings

neighbor_ratings['weighted_rating'] = (
    neighbor_ratings['rating'] * neighbor_ratings['similarity']
)

print("Ratings from Similar Users:")
display(neighbor_ratings.head(10))
```

```python
# Calculate recommendation scores

recommendation_scores = neighbor_ratings.groupby('movieId').apply(
    lambda x: x['weighted_rating'].sum() / x['similarity'].sum(),
    include_groups=False
)

recommendation_scores = recommendation_scores.sort_values(
    ascending=False
)

# Find movies already rated by the selected user

watched_movies = ratings_clean[
    ratings_clean['userId'] == target_user_id
]['movieId']

# Remove movies already rated

recommendation_scores = recommendation_scores[
    ~recommendation_scores.index.isin(watched_movies)
]

# Select top 10 recommendations

recommended_movies = recommendation_scores.head(10).reset_index()

recommended_movies.columns = ['movieId', 'predicted_rating']

# Add movie titles

recommended_movies = recommended_movies.merge(
    movies_clean[['movieId', 'title']],
    on='movieId'
)

print("Top 10 Movie Recommendations for User", target_user_id)
display(recommended_movies[['title', 'predicted_rating']])
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=recommended_movies,
    x='predicted_rating',
    y='title'
)

plt.title("Top 10 Recommended Movies for User 414")
plt.xlabel("Predicted Rating")
plt.ylabel("Movie Title")
plt.xlim(0, 5)
plt.show()
```
